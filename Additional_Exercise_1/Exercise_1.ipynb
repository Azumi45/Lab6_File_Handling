```python
# Student-Specific Information

from pathlib import Path

LAST_NAME = "Celo"
STUDENT_ID = "TUPM-26-1941"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
```

```python
# Workspace

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)
BACKUP_DIR.mkdir(exist_ok=True)

dataset_path = Path (
    r"C:\Users\admin\Lab6_File_Handling_Additional_Exercises\Additional_Exercise_1\weblog.csv"
)
```

```python
# Records Processed
sample_size = 1000 + ID_SUM

# Target Status Class
target_class = str((SEED_DIGIT % 4) + 2)

# Records Processed
records_processed = 0

# Records in Target Class
valid_records = 0

# Count per Status Code
status_counts = {}
url_counts = {}
filtered_lines = []

with open(dataset_path, "r", encoding="utf-8") as file:
    header_line = file.readline().strip().split(",")
    ip_idx, time_idx, url_idx, status_idx = 0, 1, 2, 3

    for _ in range(sample_size):
        line = file.readline()
        if not line:
            break

        records_processed += 1
        fields = line.strip().split(",")

        if len(fields) > 3:
            status_code = fields[status_idx].strip()
            if status_code.startswith(target_class):
                valid_records += 1
                status_counts[status_code] = status_counts.get(status_code, 0) + 1
                filtered_lines.append(line)

url_lines = filtered_lines
for line in url_lines:
    fields = line.strip().split(",")
    if len(fields) > url_idx:
        url_string = fields[url_idx].strip()
        url_counts[url_string] = url_counts.get(url_string, 0) + 1

# Five Most Requested URLs
top_urls_sorted = sorted(url_counts.items(), key=lambda x: x[1], reverse=True)[:5]
top_5_urls = [item[0] for item in top_urls_sorted]

print("Records processed:", records_processed)
print("Target status class:", target_class + "xx")
print("Records in target class:", valid_records)
print("Count per status code:", status_counts)

print("\nFive Most Requested URLs:")
printed_urls = set()
for line in url_lines:
    fields = line.strip().split(",")
    if len(fields) > 3:
        url_string = fields[url_idx].strip()
        
        if url_string in top_5_urls and url_string not in printed_urls:
            ip_address = fields[ip_idx].strip()
            timestamp = fields[time_idx].strip()
            status_code = fields[status_idx].strip()
            
            print(f"IP: {ip_address}, Time: {timestamp}, URL: {url_string}, Status: {status_code}")
            printed_urls.add(url_string)
```

```python
# CSV File

filtered_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_filtered_weblog.csv"

with open(filtered_path, "w", encoding="utf-8") as output:
    output.write("IP,Time,URL,Staus\n")
    for line in filtered_lines:
        output.write(line)

print("Filtered file:", filtered_path)
```

```python
# JSON Summary File

import json

json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_weblog_summary.json"

summary_data = {
    "records_processed": records_processed,
    "target_status_class": target_class + "xx",
    "records_in_target_class": valid_records,
    "status_code_counts": status_counts,
    "top_5_urls": dict(top_urls_sorted)
}

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(summary_data, file, indent=4)

print("Created:", json_path)
```

```python
# Backup File
import shutil

backup_path = BACKUP_DIR / json_path.name
shutil.copy2(json_path, backup_path)

print("Original:", json_path)
print("Backup:", backup_path)
print("Backup exists:", backup_path.exists())
```
