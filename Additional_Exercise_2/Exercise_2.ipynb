```python
# Student-Specific Information

from pathlib import Path

LAST_NAME = "Celo"
STUDENT_ID = "TUPM-26-1941"

SEED_DIGIT = int(STUDENT_ID[-1])

# Operating System
if SEED_DIGIT % 2 == 0:
    target_os = "Android"
else:
    target_os = "iOS"

# User Behavior Class
target_behavior_class = str((SEED_DIGIT % 5) + 1)

print("Operating System:", target_os)
print("User Behavior Class:", target_behavior_class)

```

```python
# Workspace

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)
BACKUP_DIR.mkdir(exist_ok=True)

dataset_path = Path (
    r"C:\Users\admin\Lab6_File_Handling_Additional_Exercises\Additional_Exercise_2\user_behavior_dataset.csv"
)
```

```python
# Total records in dataset
total_records = 0

# Records matching operating system
matching_os_count = 0

# Records in target behavior class
valid_records = 0

# Average battery drain
total_battery_drain = 0.0

# Average data usage
total_data_usage = 0.0
filtered_lines = []

# From the file; fields[2] = Operating System | fields[5] = Battery Drain (mAh/day) | fields[7] = Data Usage (MB/day) | fields[10] = User Behavior Class

with open(dataset_path, "r", encoding="utf-8") as file:
    next(file)

    for line in file:
        if not line.strip():
            continue

        total_records += 1
        fields = line.strip().split(",")

        if len(fields) > 10:
            current_os = fields[2].strip()
            current_class = fields[10].strip()

            if current_os.lower() == target_os.lower():
                matching_os_count += 1

                if int(float(current_class)) == int(target_behavior_class):
                    valid_records += 1
                    total_battery_drain += float(fields[5].strip())
                    total_data_usage += float(fields[7].strip())
                    filtered_lines.append(line)

average_battery = total_battery_drain / valid_records if valid_records else 0
average_data = total_data_usage / valid_records if valid_records else 0

print("Total records in dataset:", total_records)
print("Records matching operating system:", matching_os_count)
print("Records in target behavior class:", valid_records)
print("Average battery drain (mAh/day):", round(average_battery, 3))
print("Average data usage (MB/day):", round(average_data, 3))

# Count verification
print("Count verification:", "VERIFIED" if len(filtered_lines) == valid_records else "CHECK")
```

```python
# Exported CSV File

filtered_csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_filtered_usage.csv"

with open(filtered_csv_path, "w", encoding="utf-8") as output:
    output.write("User ID,Device Model,Operating System,Battery Drain (mAh/day),Data Usage (MB/day)\n")
    for line in filtered_lines:
        fields = line.strip().split(",")
        output.write(f"{fields[2]},{fields[5]},{fields[7]},{fields[10]}\n")

print("Exported CSV file:", filtered_csv_path)
```

```python
# JSON Summary File

import json

json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_usage_summary.json"

summary_data = {
    "operating_system": target_os,
    "user_behavior_class": int(target_behavior_class),
    "total_records_in_dataset": total_records,
    "records_matching_operating_system": matching_os_count,
    "records_in_target_behavior_class": valid_records,
    "average_battery_drain": round(average_battery, 3),
    "average_data_usage": round(average_data, 3)
}

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(summary_data, file, indent=4)

print("JSON Summary File:", json_path)
```

```python
# Backup File and Size

import shutil

file_info = filtered_csv_path.stat()

backup_path = BACKUP_DIR / filtered_csv_path.name
shutil.copy2(filtered_csv_path, backup_path)

print("Backup File:", backup_path)
print("File size:", file_info.st_size, "bytes")
```
